```python
import seaborn as sns
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd

import warnings
warnings.filterwarnings('ignore')
```

```python
df=sns.load_dataset('mpg')
```

```python
df
```

```python
df.head(10)
```

```python
df.drop('name',axis=1,inplace=True)
```

```python
df.isnull().sum()
```

```python
df.fillna(df.horsepower.median(),inplace=True)
```

```python
df.isna().sum()
```

```python
df.dtypes
```

```python
df.origin.value_counts
```

```python
df['origin']=df['origin'].map({'usa':1,'japan':2,'europe':3})
df['origin']=df['origin'].astype(int)
```

```python
df.dtypes
```

```python
X=df.drop('mpg',axis=1)
```

```python
y=df['mpg']
```

```python
X
```

```python
y
```

```python
from sklearn.model_selection import train_test_split
```

```python
 X_train, X_test, y_train, y_test =train_test_split(X,y,test_size=0.3,random_state=1)
```

```python
from sklearn.linear_model import LinearRegression
```

```python
regression_model=LinearRegression()
```

```python
regression_model
```

```python
regression_model.fit(X_train,y_train)
```

```python
X_train.columns
```

```python
for i,col_name in enumerate(X_train.columns):
        print(f"the coffecient of{col_name} is {regression_model.coef_[i]}")
```

```python
y_pred=regression_model.predict(X_test)
```

```python
from sklearn.metrics import r2_score
r2_sc=r2_score(y_test,y_pred)
print(f"the variance score is{r2_sc}")
```

```python
from sklearn.linear_model import Ridge
redge_regression_model=Ridge(alpha=0.1)
redge_regression_model.fit(X_train,y_train)

for i,col_name in enumerate(X_train.columns):
        print(f"the coffecient for {col_name} redge_regresion_model {redge_regression_model.coef_[i]}")
```

```python
y_pred=redge_regression_model.predict(X_test)
from sklearn.metrics import r2_score
r2_sc=r2_score(y_test,y_pred)
print(f"the variance score is{r2_sc}")
```

```python
from sklearn.linear_model import Lasso
lasso_regression_model=Lasso(alpha=0.5)
lasso_regression_model.fit(X_train,y_train)
```

```python
for i,col_name in enumerate(X_train.columns):
        print(f"the coffecient for {col_name} redge_regresion_model {lasso_regression_model.coef_[i]}")
```

```python
y_pred=lasso_regression_model.predict(X_test)
r2_sc=r2_score(y_test,y_pred)
print(f"the r square score is {r2_sc}")
```

```python
elastic_regression_model.fit(X_train,y_train)
```

```python
for i,col_name in enumerate(X_train.columns):
        print(f"the coffecient for {col_name} redge_regresion_model {elastic_regression_model.coef_[i]}")
```

```python
y_pred=elastic_regression_model.predict(X_test)
r2_sc=r2_score(y_test,y_pred)
print(f"the r square score is {r2_sc}")
```
